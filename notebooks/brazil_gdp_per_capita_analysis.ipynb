# PIB per Capita no Brasil — Análise Territorial com Python

**Pergunta central:** como o PIB per capita varia entre as UFs brasileiras entre 2013 e 2016 e como essas diferenças se comportam em relação ao valor nacional?


```python
import pandas as pd
df = pd.read_csv('../data/raw/pib_per_capita_brasil_2013_2016.csv')
df.head()
```

## Diagnóstico

```python
print(df.shape)
print(df.isna().sum())
print(df.duplicated(['Territorialidades','Ano']).sum())
```

## Ranking 2016

```python
ufs=df[df['Territorialidades']!='Brasil']
ufs[ufs['Ano']==2016].sort_values('PIB per capita',ascending=False).head(10)
```

## Variação 2013–2016

```python
wide=ufs.pivot(index='Territorialidades',columns='Ano',values='PIB per capita')
wide['variacao_pct']=(wide[2016]/wide[2013]-1)*100
wide['variacao_pct'].sort_values(ascending=False).round(2)
```

## Síntese

- Brasil: 22,20 em 2013 e 20,19 em 2016.
- 8 UFs acima do Brasil em 2016.
- Distrito Federal lidera; Maranhão tem o menor valor.
- Mato Grosso apresenta a maior alta; Espírito Santo, a maior queda.
